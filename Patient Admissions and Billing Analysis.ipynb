### Data Cleaning & Preprocessing

##### Check Missing Or Null Values

```python
missing_values = df.isnull().sum()  

print("Missing values in each column:",)
print(missing_values)

total_missing = df.isnull().sum().sum()
if total_missing == 0:
    print("No missing values found. No cleaning required.")
else:
    print("Missing values found. Investigate before cleaning.")
```

##### Standardize Column Names (snake_case)

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
df = df.rename(columns={"date_of_admission": "admission_date"})
print(list(df.columns))
```

##### Identify Date Columns 

```python
# Check the data type of every column
print("Current column data types:")
print(df.dtypes)
```

##### Cast Date Columns to Datetime

```python
# Convert admission dates to datetime
df["admission_date"]  = pd.to_datetime(
    df["admission_date"] 
)

# Convert discharge dates to datetime
df["discharge_date"] = pd.to_datetime(
    df["discharge_date"]
)
```

##### Validate Updated Dtypes


```python

print("Admission date type:", df["admission_date"].dtype)

print("\nDischarge date type:", df["discharge_date"].dtype)
```

##### Check & View Duplicate Records

```python
duplicate_count = df.duplicated().sum()
print("Exact duplicate rows:", duplicate_count)

duplicate_rows = df[df.duplicated()]
duplicate_rows.head(10)
```

##### Drop Duplicate Records

```python
# Save the original row count for comparison
rows_before = len(df)

# Remove exact duplicate rows
df = df.drop_duplicates()

# Count rows after cleaning
rows_after = len(df)

print("Rows removed:", rows_before - rows_after,"\nRemaining duplicates:", df.duplicated().sum())
```

##### Identify Inconsistent Text Values



```python
# Check text columns
text_columns = df.select_dtypes(include=["object", "str"]).columns
print(f"Text columns:\n{text_columns.tolist()}")

# Inspect examples of patient names
print(f"\nFirst 10 patient names:\n{df["name"].head(10).to_list()}")

# Count unique values in important categories
category_columns = [
    "gender",
    "blood_type",
    "medical_condition",
    "admission_type",
    "insurance_provider",
    "medication",
    "test_results"
]

for column in category_columns:
    print(f"\n{column}:{sorted(df[column].dropna().astype(str).unique())}")
```

##### Standardize Categorical Values



```python
# Remove spaces if exist at the beginning and end of text values
for column in text_columns:
    df[column] = df[column].str.strip()

# Standardize capitalization in selected categories
columns =  [
    "name",
    "gender",
    "blood_type",
    "medical_condition",
    "admission_type",
    "insurance_provider",
    "medication",
    "test_results"
]
for col in columns:
    df[col] = df[col].str.title()
```

##### Identify Inconsistent Age Values




```python
# Find missing age values
age_data = (df["age"].isnull().sum())
invalid_age = ((df["age"] < 0) | (df["age"] > 120)).sum()
print("Missing ages:",age_data,
      "\nMinimum age:", df["age"].min(),
      "\nMaximum age:", df["age"].max(),
      "\nAges outside screening range:", invalid_age) 
```

##### Identify Inconsistent Billing Amount



```python
# Review basic billing statistics
print(f"Billing Amount summary:\n{df["billing_amount"].describe()}")

# Count missing billing values
print(f"\nMissing billing values:{df["billing_amount"].isnull().sum()}")

# Count zero and negative bills
print(f"\nZero bills:{(df["billing_amount"] == 0).sum()}")
print(f"\nNegative bills:{(df["billing_amount"] < 0).sum()}")
```

##### Standardize Billing Amounts



```python
# Create a column as a flag for zero or negative bills
df["billing_review_needed"] = df["billing_amount"] <= 0

# Count records requiring review
print("Bills requiring review:", df["billing_review_needed"].sum())

# Display flagged records
display(df.loc[df["billing_review_needed"], ["hospital", "medical_condition", "billing_amount"]].head(10))

df.shape
```

### Export Processed Dataset

### Essential Business Questions

##### Question 1: Calculate the average billing amount by medical condition

```python
billing_by_condition = (
    df.groupby("medical_condition")["billing_amount"]
    .mean()
    .round(2)
    .sort_values(ascending=False)
)

display(billing_by_condition)
```

##### Question2: Which admission type has the highest average billing amount?

```python
billing_by_admission = (
    df.groupby("admission_type")["billing_amount"]
    .mean()
    .round(2)
    .sort_values(ascending=False)
)

display(billing_by_admission)
```

##### Question3: How many records belong to each admission type, and what percentage of the dataset does each type represent?

```python
# Count records by admission type
admission_counts = df["admission_type"].value_counts()

# Calculate percentage for each admission type
admission_percentages = (
    df["admission_type"].value_counts(normalize=True) * 100
).round(2)

# Combine results into one table
admission_summary = pd.DataFrame({
    "Record Count": admission_counts,
    "Percentage (%)": admission_percentages
})

display(admission_summary)
```
